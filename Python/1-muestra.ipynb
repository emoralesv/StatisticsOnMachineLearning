```python
import math

# Datos del problema
z = 1.64 # valor crítico para 90%
varianza = 60
sigma = math.sqrt(varianza)
magnitud_error = 4
d = magnitud_error/2.0  # margen de error (semi-ancho)

# Fórmula del tamaño de muestra
n = (z * sigma / d)**2
n_redondeado = math.ceil(n)

print("n calculado:", n)
print("n requerido (redondeado):", n_redondeado)
```
