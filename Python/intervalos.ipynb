```python
import numpy as np
import pandas as pd
import math

# ---------------------------
# Ejemplo: Altura de 100 estudiantes (en metros)
# Sustituye esta lista por tus datos reales
# ---------------------------
np.random.seed(42)  # para reproducibilidad
N = 100
datos = np.round(np.random.normal(loc=1.65, scale=0.1, size=N), 2)
# Creamos DataFrame con índice de sujeto y su altura
df = pd.DataFrame({
    "Sujeto n": range(1, N+1),   # del 1 al 100
    "Altura (m)": datos
})
df = df.set_index("Sujeto n")

frecuencia_abs = df["Altura (m)"].value_counts().sort_index()

# Convertimos a DataFrame ordenado
df_freq = pd.DataFrame({
    "Altura (m)": frecuencia_abs.index,
    "Frecuencia absoluta (f)": frecuencia_abs.values
})
df_freq.set_index("Altura (m)")

```

```python

print(df.describe())
```

```python
n = len(df)
k = round(1 + 3.322 * math.log10(n))
# Construcción de intervalos
intervalos = pd.interval_range(start=df["Altura (m)"].min(), 
                               end=df["Altura (m)"].max(), 
                               periods=k)
print("Número de clases (Sturges):", k)

```

```python
# Cálculo de frecuencias absolutas por intervalo
frecuencia_abs = pd.cut(df["Altura (m)"], bins=intervalos).value_counts().sort_index()

# Crear tabla de distribución de frecuencias
tabla_intervalos = pd.DataFrame({
    "Clase": range(1, len(frecuencia_abs) + 1),
    "Intervalo": frecuencia_abs.index.astype(str),
    "f": frecuencia_abs.values
})

# Frecuencia relativa
tabla_intervalos["fr"] = tabla_intervalos["f"] / n

# Frecuencia acumulada
tabla_intervalos["F"] = tabla_intervalos["f"].cumsum()

# Frecuencia relativa acumulada
tabla_intervalos["Fr"] = tabla_intervalos["f"].cumsum()

# Hacer que Clase sea el índice
tabla_intervalos = tabla_intervalos.set_index("Clase")
print(tabla_intervalos)
```
